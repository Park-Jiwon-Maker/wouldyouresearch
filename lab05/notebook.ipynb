# 직접 해보기 05 · 인산염의 지문 찾기

인산염 논문 저자들이 공개한 **실험실** 질량 스펙트럼(Zenodo 10.5281/zenodo.7703848, CC BY 4.0)을 읽는다. 카시니가 토성에서 잰 자료가 아니다.

같은 계산이 `analysis.py`에도 들어 있다.

```python
import analysis as A
rows = {k: A.load(k) for k in A.SOLUTIONS}
for k, r in rows.items():
    print(A.SOLUTIONS[k], A.sha8(k), len(r), '행')
```

## 1. 전체 구간(20–210)과 확대 구간(110–200)을 그려 본다
matplotlib이 없으면 `pip install matplotlib`.

```python
import matplotlib.pyplot as plt
fig, axes = plt.subplots(2, 1, figsize=(9, 6))
for ax, (lo, hi) in zip(axes, [(20, 210), (110, 200)]):
    sel = [(m, v) for m, v in rows['mix'] if lo <= m <= hi]
    top = max(v for _, v in sel)
    ax.plot([m for m, _ in sel], [v / top for _, v in sel], lw=.6, color='k')
    for c in A.MARKS: ax.axvspan(c - 1, c + 1, color='teal', alpha=.15)
    ax.set_xlim(lo, hi); ax.set_ylabel('상대 신호')
axes[1].set_xlabel('질량/전하 (u/e)'); plt.tight_layout()
```

## 2. 네 용액의 봉우리 높이를 비교한다 (책의 예상 결과)

```python
for k in A.SOLUTIONS:
    r = A.analyse(rows[k], window=1.0)
    print(f"{A.SOLUTIONS[k]:<22}", {c: p['height'] for c, p in r['peaks'].items()})
```
